```python
from pm4py.objects.petri_net.obj import PetriNet, Marking
from pm4py.objects.petri_net.utils import petri_utils

# Create a simple ideal Petri net
net = PetriNet("Ideal_Discharge_Flow")

# Define places
p1 = PetriNet.Place("p1")
p2 = PetriNet.Place("p2")
p3 = PetriNet.Place("p3")
p4 = PetriNet.Place("p4")
p5 = PetriNet.Place("p5")
p6 = PetriNet.Place("p6")

# Define transitions (activities)
t1 = PetriNet.Transition("Discharge Order Written", "Discharge Order Written")
t2 = PetriNet.Transition("Pharmacy Request Sent", "Pharmacy Request Sent")
t3 = PetriNet.Transition("Pharmacy Prep Start", "Pharmacy Prep Start")
t4 = PetriNet.Transition("Pharmacy Prep Complete", "Pharmacy Prep Complete")
t5 = PetriNet.Transition("Final Physician Sign-off", "Final Physician Sign-off")
t6 = PetriNet.Transition("Patient Leaves Facility", "Patient Leaves Facility")

# Add all to the net
net.places.update({p1, p2, p3, p4, p5, p6})
net.transitions.update({t1, t2, t3, t4, t5, t6})

# Connect places and transitions
petri_utils.add_arc_from_to(p1, t1, net)
petri_utils.add_arc_from_to(t1, p2, net)
petri_utils.add_arc_from_to(p2, t2, net)
petri_utils.add_arc_from_to(t2, p3, net)
petri_utils.add_arc_from_to(p3, t3, net)
petri_utils.add_arc_from_to(t3, p4, net)
petri_utils.add_arc_from_to(p4, t4, net)
petri_utils.add_arc_from_to(t4, p5, net)
petri_utils.add_arc_from_to(p5, t5, net)
petri_utils.add_arc_from_to(t5, p6, net)
petri_utils.add_arc_from_to(p6, t6, net)

# Define markings
im = Marking()
fm = Marking()
im[p1] = 1  # start place
fm[p6] = 1  # end place

print("✅ Ideal TO-BE model created successfully.")
```

```python
from pm4py.objects.petri_net.obj import PetriNet, Marking
from pm4py.objects.petri_net.utils import petri_utils

# Create ideal Petri net
net = PetriNet("Ideal_Discharge_Flow")

# Define places
p1 = PetriNet.Place("p1")
p2 = PetriNet.Place("p2")
p3 = PetriNet.Place("p3")
p4 = PetriNet.Place("p4")
p5 = PetriNet.Place("p5")
p6 = PetriNet.Place("p6")

# Define transitions (activities)
t1 = PetriNet.Transition("Discharge Order Written", "Discharge Order Written")
t2 = PetriNet.Transition("Pharmacy Request Sent", "Pharmacy Request Sent")
t3 = PetriNet.Transition("Pharmacy Prep Start", "Pharmacy Prep Start")
t4 = PetriNet.Transition("Pharmacy Prep Complete", "Pharmacy Prep Complete")
t5 = PetriNet.Transition("Final Physician Sign-off", "Final Physician Sign-off")
t6 = PetriNet.Transition("Patient Leaves Facility", "Patient Leaves Facility")

# Add to net
for p in [p1, p2, p3, p4, p5, p6]:
    net.places.add(p)
for t in [t1, t2, t3, t4, t5, t6]:
    net.transitions.add(t)

# Connect places and transitions
petri_utils.add_arc_from_to(p1, t1, net)
petri_utils.add_arc_from_to(t1, p2, net)
petri_utils.add_arc_from_to(p2, t2, net)
petri_utils.add_arc_from_to(t2, p3, net)
petri_utils.add_arc_from_to(p3, t3, net)
petri_utils.add_arc_from_to(t3, p4, net)
petri_utils.add_arc_from_to(p4, t4, net)
petri_utils.add_arc_from_to(t4, p5, net)
petri_utils.add_arc_from_to(p5, t5, net)
petri_utils.add_arc_from_to(t5, p6, net)
petri_utils.add_arc_from_to(p6, t6, net)

# Define initial and final markings
im = Marking()
fm = Marking()
im[p1] = 1
fm[p6] = 1

print("✅ Ideal TO-BE model created successfully.")
```
